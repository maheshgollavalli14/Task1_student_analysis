```python
import pandas as pd
import matplotlib.pyplot as plt
df=pd.read_csv(r"C:\Users\Lenovo\Downloads\student\student-mat.csv",sep=';')
print(df.columns)



```

```python
print(df.head())
print("--------------------------------------------------------")
print(df.shape)
```

```python
print(df.dtypes)
```

```python
print(df.info())
```

```python
#checking the missing values
print(df.isnull().sum())
```

```python
#removing the duplicates
df=df.drop_duplicates()
#avg the final grade (G3)
print("Avg of final grade(G3)=",df['G3'].mean())
```

```python
#students who got above 15 marks
c=(df['G3']>15).sum()
print("Students got above 15 marks is : ",c)
```

```python
#study time correlated with performance
corr=df['studytime'].corr(df['G3'])
print("Correlation : ",corr)
```

```python
#which gender performs better
print(df.groupby('sex')['G3'].mean())
```

```python
#visualization
#Histogram of grades:

plt.hist(df['G3'],bins=10,color='lightgreen',edgecolor='black')
plt.title("DISTRIBUTION OF FINAL GRADES")
plt.xlabel("GRADES")
plt.ylabel("FREQUENCY")
plt.show()
```

```python
#Scatterplot: study time vs grades.
plt.scatter(df['studytime'],df['G3'])
plt.title("study time vs grades")
plt.ylabel("studytime")
plt.xlabel("grade")
plt.show()
```

```python
#Bar chart: male vs female average score.
avg=df.groupby('sex')['G3'].mean()
avg.plot(kind='bar')
plt.title("male vs female avg score")
plt.ylabel("avg")
plt.xlabel("gender")
plt.show()
```
